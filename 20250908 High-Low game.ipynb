```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ntrials=10000000
```

```python
rannos=np.random.random((ntrials,3))
```

```python
win1 = (((rannos[:,0]>=0.5)&(rannos[:,1]<rannos[:,0]))|((rannos[:,0]<0.5)&(rannos[:,1]>rannos[:,0])))
```

```python
win2 = win1&(((rannos[:,1]>=0.5)&(rannos[:,2]<rannos[:,1]))|((rannos[:,1]<0.5)&(rannos[:,2]>rannos[:,1])))
```

```python
plt.plot([round(i.item(),2) for i in np.arange(0,1,0.01)], [np.mean(win1[(rannos[:,0]>=i)&(rannos[:,0]<i+0.01)]).item() for i in np.arange(0,1,0.01)])
plt.show()
```

```python
plt.hist(rannos[:,1][win1], bins=100)
plt.show()
```

```python
plt.plot([round(i.item(),2) for i in np.arange(0,1,0.01)], [np.mean(win2[(rannos[:,0]>=i)&(rannos[:,0]<i+0.01)]).item() for i in np.arange(0,1,0.01)])
plt.show()
```

```python
plt.hist(rannos[:,1][win2], bins=100)
plt.show()
```

```python
for i in np.arange(0,1,0.01):
    print(round(i,2),np.mean(win2[(rannos[:,0]>=i)&(rannos[:,0]<i+0.01)]))
```
