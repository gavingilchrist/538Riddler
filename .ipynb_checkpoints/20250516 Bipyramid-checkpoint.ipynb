```python
import numpy as np

cells = np.arange(16).reshape(4, 4)
cell_links = {i: list() for i in range(15)}
for f, t in np.concatenate((np.column_stack((cells[:-1].reshape(-1), 
                                             cells[1:].reshape(-1))),
                            np.column_stack((cells[:, :-1].reshape(-1), 
                                             cells[:, 1:].reshape(-1))),
                            np.column_stack((cells[:-1, 1:].reshape(-1), 
                                             cells[1:, :-1].reshape(-1))),
                            np.column_stack((cells[1:, :-1].reshape(-1), 
                                             cells[:-1, 1:].reshape(-1)))),
                           axis=0):
    cell_links[f] += [int(t)]

def get_n_paths(cell=0, visited=None):
    visited = visited or [0]
    if cell == len(cell_links):
        return 1
    else:
        return sum([get_n_paths(c, [*visited, c])
                    for c in cell_links[cell] if c not in visited])

get_n_paths()
```

```python
# Extra credit
vlev = [[[0., 0.]]]
for i in range(3):
    vlev += [[[float(x + (np.sin(a) / np.sqrt(3))), 
               float(y + (np.cos(a) / np.sqrt(3)))]
              for x, y in vlev[-1]
              for a in [a*2/3*np.pi for a in range(3)]]]
vlev = [[[round(j, 6) for j in v[i]]
         for i in range(len(v))
         if all([any([round(v[j][k], 6) != round(v[i][k], 6) 
                      for k in range(2)]) 
                 for j in range(i)])]
        for v in vlev]
vlev += vlev[:-1][::-1]
vertices = np.array([[x, y, round(ln*((2/3)**0.5), 6)]
                     for ln, l in enumerate(vlev)
                     for x, y in l])

filt = ((np.round(np.sum((np.expand_dims(vertices, 1) 
                          - np.expand_dims(vertices, 0)) ** 2, axis=2), 
                  4) == 1)
        & (np.round(np.expand_dims(vertices[:, 2], 0) 
                    - np.expand_dims(vertices[:, 2], 1), 
                    4) >= 0))
cell_links = {i: np.arange(vertices.shape[0])[filt[i]].tolist() 
              for i in range(vertices.shape[0])
              if np.any(filt[i])}

known = dict()
def get_n_paths(cell=0, visited=None):
    if visited is None:
        visited = np.zeros(np.sum(vertices[:, 2] == vertices[cell, 2]), dtype=bool)
    dkey = ('%02d' % cell) + '|'+ (''.join([str(int(i)) for i in visited]))
    if dkey not in known:
        if cell == len(cell_links):
            known[dkey] = 1
        else:
            lev_cell_ix = np.arange(vertices.shape[0])[vertices[:, 2] == vertices[cell, 2]]
            known[dkey] = sum([get_n_paths(c, np.zeros(np.sum(vertices[:, 2] == vertices[c, 2]), dtype=bool))
                               if vertices[c, 2] > vertices[cell, 2]
                               else get_n_paths(c, np.where(lev_cell_ix == cell, True, visited))
                               for c in cell_links[cell]
                               if np.all(lev_cell_ix[visited] != c)])

    return known[dkey]

get_n_paths()
```
