```python
import numpy as np
import matplotlib.pyplot as plt

holes = np.column_stack(([j for i in range(7) for j in [i*(0.75**0.5)]*(7-i)],
                         [j+(0.5*i) for i in range(7) for j in range(7-i)]))
ix = np.array([[i, j, k]
               for i in range(26) 
               for j in range(i+1, 27) 
               for k in range(j+1, 28)])
dist = np.column_stack([np.sqrt(np.sum((holes[ix[:, i]] - holes[ix[:, j]]) ** 2, axis=1))
                        for i in range(2) for j in range(i+1, 3)])
dist = np.round(dist, 10)

eq_filt = np.ptp(dist, axis=1) == 0
np.sum(eq_filt)
```

```python
eq_ix = ix[eq_filt][np.lexsort((np.arange(np.sum(eq_filt)), dist[eq_filt, 0]))]
for eqn in np.arange(eq_ix.shape[0]):
    print(f'#{eqn+1}')
    plt.plot(holes[:, 1], holes[:, 0], 'ok', markersize=12)
    plt.plot(holes[eq_ix[eqn], 1][[0, 1, 2, 0]], holes[eq_ix[eqn], 0][[0, 1, 2, 0]], '-r')
    plt.show()
    print()
```
