```python
def get_longest(row=None):
    row = row or 'AB'
    prev_trios = [row[i:i+3] for i in range(len(row)-2)]
    next_opt = [row+i for i in 'ABC' 
                if i != row[-1] 
                and row[-2:]+i not in prev_trios]
    if len(next_opt) == 0:
        return row
    else:
        best = None
        bestlen = 0
        for r in next_opt:
            res = get_longest(r)
            if len(res) > bestlen:
                best = res
                bestlen = len(res)
        return best

result = get_longest(row=None)
print(result, len(result))
```
