```python
import numpy as np
```

```python
bearings = np.empty((0), dtype=float)
x = 0
for i in range(10000):
    x += 1
    bearings = np.unique(np.append(bearings, (np.atan(np.arange(x+1) / x) / np.pi * 180)))
    angles = bearings[1:] - bearings[:-1]
    if angles.shape[0] > 2:
        print(x, 
              '%.2f' % (np.max(angles[1:-1])), 
              ('%.2f' % (bearings[np.argmax(angles[1:-1])+1]))+'-'
               +('%.2f' % (bearings[np.argmax(angles[1:-1])+2])))
```
