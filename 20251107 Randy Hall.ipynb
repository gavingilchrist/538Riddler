```python
import matplotlib.pyplot as plt

staypct = 0.6

doors = [[1,0,0]]
for i in range(30):
    doors += [[staypct*doors[-1][0] + 0.4*doors[-1][1],
               (1-staypct)*doors[-1][0] + 0.2*doors[-1][1] + (1-staypct)*doors[-1][2],
               staypct*doors[-1][2] + 0.4*doors[-1][1]]]

for i in zip(*doors):
    plt.plot(range(31), i)
plt.show()
```
