```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_strands = 1000000
grid_gr = 100
grid_limits = np.linspace(0, 1, grid_gr+1)

pt1 = np.random.random((n_strands, 2))
pt2 = np.random.random((n_strands, 2))

sl = (pt1[:, 1:] - pt2[:, 1:]) / (pt1[:, :1] - pt2[:, :1])
yint = pt1[:, 1:] - (sl * pt1[:, :1])

vx = (np.expand_dims(grid_limits, 0) * sl) + yint
vx = np.floor(vx * grid_gr).astype(int)

hx = (np.expand_dims(grid_limits, 0) - yint) / sl
hx = np.floor(hx * grid_gr).astype(int)

xcells = []
filt = (vx[:, :-1].reshape(-1) >= 0) & (vx[:, :-1].reshape(-1) < grid_gr)
xcells += [(vx[:, :-1].reshape(-1)[filt] * grid_gr) + np.tile(np.arange(grid_gr), n_strands)[filt]]
filt = (vx[:, 1:].reshape(-1) >= 0) & (vx[:, 1:].reshape(-1) < grid_gr)
xcells += [(vx[:, 1:].reshape(-1)[filt] * grid_gr) + np.tile(np.arange(grid_gr), n_strands)[filt]]
filt = (hx[:, :-1].reshape(-1) >= 0) & (hx[:, :-1].reshape(-1) < grid_gr)
xcells += [(np.tile(np.arange(grid_gr), n_strands)[filt] * grid_gr) + hx[:, :-1].reshape(-1)[filt]]
filt = (hx[:, 1:].reshape(-1) >= 0) & (hx[:, 1:].reshape(-1) < grid_gr)
xcells += [(np.tile(np.arange(grid_gr), n_strands)[filt] * grid_gr) + hx[:, 1:].reshape(-1)[filt]]
xcells = np.concatenate(xcells)

heatmap = np.zeros((grid_gr, grid_gr), dtype=float)
v, c = np.unique(xcells, return_counts=True)
heatmap[(v / grid_gr).astype(int), np.mod(v, grid_gr)] = c / n_strands / 2

plt.imshow(heatmap, cmap='gray')
```
