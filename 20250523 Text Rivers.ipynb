```python
import numpy as np
import matplotlib.pyplot as plt

ntrials = 10000
nchrs = 1000000

cw_len = np.random.randint(3, 5, ntrials)
chr_lett = np.zeros(ntrials, dtype=int)
p_space = [float(np.mean(chr_lett == -1))]
for i in range(1, nchrs):
    chr_lett += 1
    chr_lett[chr_lett == cw_len] = -1
    cw_len[chr_lett == -1] = np.random.randint(3, 5, np.sum(chr_lett == -1))
    p_space += [float(np.mean(chr_lett == -1))]

plt.plot(np.arange(100), p_space[-100:])
plt.show()

print(np.mean(p_space[-100:]))
```
