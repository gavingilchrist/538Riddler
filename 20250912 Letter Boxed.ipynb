```python
import numpy as np
```

```python
poss = np.empty((1,0), dtype='<U1')
for i in range(8):
    poss = np.column_stack((np.repeat(poss, 8, axis=0),
                            np.tile([*'ABCDEFGH'], poss.shape[0])))
    if i:
        poss = poss[np.all((poss[:, -1:] != poss[:, :-1]), axis=1) 
                    & (np.ptp((poss[:, -2:].view(np.uint32)-1)//2, axis=1) > 0)]
poss.shape
```

```python
known={}
def get_n_poss(edges=None, last=-1):
    edges = edges or [2, 2, 2, 2]
    dkey = ''.join([str(i) for i in edges]) + '|' + ('X' if last < 0 else str(last))
    if dkey not in known:
        if sum(edges) == 0:
            known[dkey] = 1
        else:
            known[dkey] = sum([get_n_poss([k-(j==i) for j,k in enumerate(edges)], i) 
                               for i in range(4) 
                               if edges[i] and i != last])
    return known[dkey]
```
