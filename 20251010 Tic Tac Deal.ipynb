```python
import numpy as np

# Get all dice possibilities, resulting totals for each toss
dice = np.column_stack([np.arange(6**6)%6**(i+1)//6**i+1 for i in range(6)][::-1])
dice_totals = np.column_stack([np.sum(dice[:,i:i+2], axis=1) for i in range(0,6,2)])

# Get boolean flags
# Between 3 and 11
c1 = (np.min(dice_totals, axis=1) > 2) & (np.max(dice_totals, axis=1) < 12)
# All different
c2 = np.all(np.column_stack([dice_totals[:, i]!=dice_totals[:, j]for i,j in[[0,1],[0,2],[1,2]]]), axis=1)
# Same row
c3 = (np.ptp(dice_totals//3 - 1, axis=1) == 0)
# Same column
c4 = (np.ptp(dice_totals%3, axis=1) == 0)
# NE-SW Diagonal
c5 = (np.ptp(dice_totals//3+dice_totals%3, axis=1) == 0)
# NW-SE Diagonal
c6 = (np.ptp(dice_totals//3-dice_totals%3, axis=1) == 0)

# Calculate total probability of line
np.mean(c1&c2&(c3|c4|c5|c6))
```
