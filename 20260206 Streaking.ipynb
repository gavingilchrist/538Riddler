```python
import numpy as np

poss = np.array([[a,b,c,d,e,f]
                 for a in range(1,10)
                 for b in range(a,10)
                 for c in range(b,10)
                 for d in range(c,10)
                 for e in range(d,10)
                 for f in range(5)])

def get_mode(arr):
    numcts = np.column_stack(([np.sum(arr==i, axis=1) for i in range(1,10)]))
    return np.where(np.sum((z:=numcts == np.max(numcts, axis=1, keepdims=True)), 
                           axis=1) == 1,
                    np.argmax(z, axis=1)+1,
                    10)
    
ages_pre = poss[:,:5]
mean_pre = np.mean(ages_pre, axis=1)
med_pre = np.median(ages_pre, axis=1)
mode_pre = get_mode(ages_pre)
range_pre = np.ptp(ages_pre, axis=1)

ages_post = poss[:,:5] + (np.expand_dims(np.arange(5),0) == poss[:,5:])
med_post = np.median(ages_post, axis=1)
mode_post = get_mode(ages_post)

filt = ((np.ptp(np.column_stack((mean_pre, med_pre, mode_pre, range_pre)), axis=1) == 0)
        & (med_post == mode_post) & (med_pre != med_post))
poss[filt]


# Before, kids ages 2,5,5,6,7
#     Mean = 5
#     Median = 5
#     Mode = 5
#     Range = 5

# After, kids ages 2,5,6,6,7
#     Mean = 5.2
#     Median = 6
#     Mode = 6
#     Range = 5
```
