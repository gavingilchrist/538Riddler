```python
def get_equal_sets(primes,nsets,targ=None):
    if targ is None:
        primes = {*primes}
        targ = sum(primes) // nsets
    if nsets:
        return [[sorted(i)]+j
                for i in get_set(primes, targ)
                for j in get_equal_sets(primes-{*i},nsets-1,targ)]
    else:
        return [[]]
    
def get_set(x,y,z=None):
    if y==0:
        return [[]]
    elif z is None:
        return [[i]+j
                for i in [max(x)]
                for j in get_set(x-{i},y-i,i)]
    else:
        return [[i]+j
                for i in x
                if i <= min(y,z or y)
                for j in get_set(x-{i},y-i,i)]
```

```python
nsets = 3

primes = [2]
while True:
    if sum(primes)%nsets == 0 and (eqsets:=get_equal_sets(primes,nsets)):
        break
    primes += [primes[-1]+1]
    while any([primes[-1]%i==0 for i in primes if i <= primes[-1]**0.5]):
        primes[-1] += 1

print(len(primes))
for i in eqsets:
    print(i)
```
