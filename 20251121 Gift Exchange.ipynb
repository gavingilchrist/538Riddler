```python
poss = [dict(zip(list('ABCDE'), [a,b,c,d,e]))
        for a in 'BCDE' for b in 'ACDE' for c in 'ABDE' for d in 'ABCE' for e in 'ABCD'
        if len({a,b,c,d,e}) == 5]

track_links = lambda x,y=5,z='A': [z]+track_links(x, y-1, x[z]) if y else [z]
full_circle = lambda x: len({*(z:=track_links(x))}) == 5 and z[0]==z[-1]

sum([full_circle(i) for i in poss]) / len(poss)
```

```python
import random

def do_trial():
    k = list('ABCDE')
    v = list('ABCDE')
    while any([i==j for i,j in zip(k,v)]):
        random.shuffle(v)
    return full_circle(dict(zip(k,v)))

ntrials = 1000000
sum([do_trial() for i in range(ntrials)]) / ntrials
```
