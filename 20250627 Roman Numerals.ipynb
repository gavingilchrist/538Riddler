```python
rnums = ['I', 'II', 'III', 'IV', 'V', 'VI', 'VII', 'VIII']
rn_decode = dict(zip(rnums, range(1, len(rnums)+1)))

known = dict()
def get_poss(code):
    if not code in known:
        if code == '':
            known[code] = [[]]
        else:
            known[code] = [[rn_decode[i], *j]
                           for i in rnums
                           for j in get_poss(code[len(i):])
                           if code.startswith(i)]
    return known[code]
```

```python
# Part 1
all_poss = get_poss('IIIIIIIIII')
len(all_poss)
```

```python
# Part 2
all_poss = get_poss('IIIVIIIVIIIVIII')
len(all_poss)
```
