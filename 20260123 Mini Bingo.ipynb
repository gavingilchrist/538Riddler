```python
import numpy as np

isbingo = lambda x: (np.any(np.all(x, axis=0))
                     | np.any(np.all(x, axis=1))
                     | np.all(x.reshape(-1)[::4])
                     | np.all(x.reshape(-1)[2:-2:2])).item()
```

```python
# Random trial
def trial():
    grid = np.pad([[1]], 1)
    ntokens = 0
    while not isbingo(grid):
        while grid[(y:=np.random.randint(0,3)), (x:=np.random.randint(0,3))] == 1:  pass
        grid[y,x] = 1
        ntokens += 1
    return ntokens

np.mean([trial() for i in range(100000)]).item()
```

```python
# Recursive calc
def exp_tokens(grid):
    if isbingo(grid):
        return {0: 1.}
    else:
        results = dict()
        for y in [0,1,2]:
            for x in [0,1,2]:
                if grid[y,x]==0:
                    for n,ct in exp_tokens(np.where((np.expand_dims([0,1,2],1)==y) 
                                                    & (np.expand_dims([0,1,2],0)==x), 
                                                    1, 
                                                    grid)).items():
                        results[n+1] = results.get(n+1,0)+(ct/np.sum(grid==0).item())
        return results

sum([k*v for k,v in exp_tokens(np.pad([[1]], 1)).items()])
```
