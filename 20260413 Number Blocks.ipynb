```python
blocks = np.array([['XXXXXX' for i in '0123']])
n=0
while True:
    n += 1
    
    # Identify block sets that would need to be altered to display current number
    need_alt = np.ones(blocks.shape[0], dtype=bool)
    for p in allperm(nineconv(f'{n: 04d}')):
        need_alt[need_alt] = ~np.all(np.any(((z:=np.expand_dims(p,(0,2)))==' ') 
                                            | (blocks[need_alt].view('<U1').reshape(-1,4,6) == z), 
                                            axis=2), 
                                     axis=1)
    
    if need_alt.shape[0]:
        bl_new = {*blocks[~need_alt].view('<U24')[:,0].tolist()}
        
        # For those sets, identify all possible ways to make additions
        for p in allperm(nineconv(f'{n: 04d}')):
            addnum = np.where(np.any(((y:=np.expand_dims(p,(0,2)))==' ') 
                                     | ((z:=blocks[need_alt].view('<U1').reshape(-1,4,6)) == y), 
                                     axis=2), 
                              'X', np.where(np.any(z=='X', axis=2), y[:,:,0], 'E'))
            bl_new.update({*np.sort(np.sort(np.append(z[(filt:=np.all(addnum != 'E', axis=1))], 
                                                      np.expand_dims(addnum[filt],2), 
                                                      axis=2).view(np.uint32), 
                                            axis=2).view('<U1')[:,:,:6].view('<U6')[:,:,0],
                                    axis=1).view('<U24')[:,0].tolist()})
            
        if bl_new:
            blocks = np.array(sorted(bl_new)).view('<U6').reshape(-1,4)
        else:
            n -= 1
            break
        
    print(n, blocks.shape[0])
    
n, blocks
```
