```python
import numpy as np

cells = np.arange(16).reshape(4, 4)
cell_links = {i: list() for i in range(15)}
for f, t in np.concatenate((np.column_stack((cells[:-1].reshape(-1), 
                                             cells[1:].reshape(-1))),
                            np.column_stack((cells[:, :-1].reshape(-1), 
                                             cells[:, 1:].reshape(-1))),
                            np.column_stack((cells[:-1, 1:].reshape(-1), 
                                             cells[1:, :-1].reshape(-1))),
                            np.column_stack((cells[1:, :-1].reshape(-1), 
                                             cells[:-1, 1:].reshape(-1)))),
                           axis=0):
    cell_links[f] += [t]

def get_n_paths(cell=0, visited=None):
    visited = visited or [0]
    if cell == 15:
        return 1
    else:
        return sum([get_n_paths(c, [*visited, c])
                    for c in cell_links[cell] if c not in visited])

get_n_paths()
```
